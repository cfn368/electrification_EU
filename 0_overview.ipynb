# Analysis basis

**Idea:** An approach to assess if aggregates differences in electrification stem from different industry-compositions between countries or different degrees of electrification with industries.
A framework for doing so is a *shift-share decomposition*. 

---

**Model:** I assess the aggregate electrification rate for country $c$ at time $t$ as a weighted average,
$$
    E_{c,t} = \sum_i s_{i,c,t} \cdot e_{i,c,t},
$$
where $s_{i,c,t}$ is industry $i$'s share of total energy within country $c$ at time $t$, and $e_{i,c,t}$ is the within-industry electrification rate. *Claude: Consider LMDI instead.*

**Interpretation:** Then as I compare two countries, the gap $\Delta E$ can come from two places:
1. Different industry mixes, $s$.
2. Different electrification intensities within industries, $e$.

---

**Decomposition:** Comparing country $A$ to a benchmark $B$ (EU average, a specific country, or the same country in an earlier period), the gap splits into three terms:
$$
    E_A - E_B = \underbrace{\sum_i (s_{i,A} - s_{i,B}) \cdot \bar{e}_i}_{\text{structural / between}} + \underbrace{\sum_i \bar{s}_i \cdot (e_{i,A} - e_{i,B})}_{\text{within / intensity}} + \underbrace{\sum_i (s_{i,A} - s_{i,B})(e_{i,A} - e_{i,B})}_{\text{interaction}},
$$
where $\bar{s}_i$ and $\bar{e}_i$ are the benchmark values (or averages of the two).

1. *Structural (between):* share of the gap explained by $A$ having more/less of heavy vs. light industries, holding within-industry electrification fixed.
2. *Within (intensity):* share explained by $A$ electrifying each industry more/less intensively than $B$, holding the industry mix fixed.
3. *Interaction:* residual covariance — $A$ both has more of the industries where it is also more electrified.

---

**Axes:** With EU countries across time, two natural comparisons:
1. *Cross-country at a given $t$:* decompose $E_{c,t} - E_{EU,t}$ into structure vs. intensity. E.g. is Denmark more electrified than Germany because it has less heavy industry, or because each industry is more electrified?
2. *Within-country over time:* decompose $E_{c,t} - E_{c,t-k}$. Tells whether a country's electrification progress comes from deindustrialization (structure shifting toward already-electric sectors) or from genuine within-industry electrification.
